```python
# input = "time flies"

# Transition Matrix
#       N       V
# N  [ 0.3     0.7 ]
# V  [ 0.8     0.2 ]


# Emission Matrix
#        time    flies
# N    [ 0.5      0.1 ]
# V    [ 0.2      0.6 ]
```

```python
states = ['N','V']
words =  ['time','flies']

transition = [[0.3,0.7],
              [0.8,0.2]]

emission = [[0.5,0.1],
            [0.2,0.6]]
```

```python
# Viterbi matrix
V = []
path = []

for i in range(len(states)):
  row = []
  for j in range(len(words)):
    row.append(0)
  V.append(row)
  path.append([""]*len(words))
```

```python
# Initialization
for i in range(len(states)):
  V[i][0] = emission[i][0]
  path[i][0] = states[i]
```

```python
# Recursion
for t in range(1, len(words)):
  for j in range(len(states)):
    max_prob = 0
    best_state = 0
```

```python
for i in range(len(states)):
      prob = V[i][t-1]*transition[i][j]*emission[j][t]
      if prob > max_prob:
        max_prob = prob
        best_state = i

        V[j][t] = max_prob
        path[j][t] = path[best_state][t-1] + " " + states[j]
```

```python
if V[0][1] > V[1][1]:
  best_sequence = path[0][1]
  print("Most likely POS Sequence : ", best_sequence)

else:
  best_sequence = path[1][1]
  print("Most likely POS Sequence : ", best_sequence)

print("Viterbi Matrix : ", V)
```
