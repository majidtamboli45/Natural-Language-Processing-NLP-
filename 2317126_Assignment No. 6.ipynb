### Using Lesk 

```python
import nltk
nltk.download('wordnet')
nltk.download('punkt')
nltk.download('stopwords')
```

```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

sentence1 = "He deposited money in the bank"
sentence2 = "The fisherman sat on the bank of the river"

word = "bank"

sense1 = lesk(word_tokenize(sentence1), word)
sense2 = lesk(word_tokenize(sentence2), word)

print("Sentence 1 Sense:", sense1)
print("Definition:", sense1.definition())

print("\nSentence 2 Sense:", sense2)
print("Definition:", sense2.definition())
```

### USing Overlapping

```python
import nltk
from nltk.corpus import wordnet as wn, stopwords
from nltk.tokenize import word_tokenize

stop = set(stopwords.words('english'))

def wsd(sentence, word):
    context = [w for w in word_tokenize(sentence.lower()) if w not in stop]
    best_sense, max_overlap = None, 0

    for syn in wn.synsets(word):
        signature = word_tokenize(syn.definition())
        for ex in syn.examples():
            signature += word_tokenize(ex)

        overlap = len(set(context) & set(signature))

        if overlap > max_overlap:
            best_sense, max_overlap = syn, overlap

    return best_sense


sentence = "The River over"
sense = wsd(sentence, "bank")

print("Sense:", sense.name())
print("Definition:", sense.definition())
```
