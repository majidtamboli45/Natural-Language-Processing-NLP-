### Preprocessing

```python
import nltk
import re
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
def preprocess_text(text):
    text = text.lower()                     
    text = re.sub(r'[^a-z\s]', '', text)   
    tokens = word_tokenize(text)            
    return tokens

df['tokens'] = df['reviewText'].apply(preprocess_text)
```

```python
df['tokens']
```

### Word2Vec Model

```python
from gensim.models import Word2Vec
```

```python
Word2Vec_model = Word2Vec(sentences=df['tokens'],vector_size = 100,window = 10,min_count=5,workers=4)
```

```python
Word2Vec_model.wv['good']
```

```python
Word2Vec_model.wv.most_similar('good')
```

```python
Word2Vec_model.wv.most_similar('bad')
```

```python
similarity_score = Word2Vec_model.wv.similarity("good", "bad")
print("Similarity Score is : ", similarity_score)
```

```python
similarity_score = Word2Vec_model.wv.similarity("good", "great")
print("Similarity Score is : ", similarity_score)
```

### FastText Model

```python
from gensim.models import FastText
```

```python
fasttext_model = FastText(sentences=df['tokens'],vector_size=100,window=5,min_count=1,workers=4)
```

```python
fasttext_model.wv['good']
```

```python
fasttext_model.wv.most_similar('good')
```

```python
fasttext_model.wv.most_similar('bad')
```

```python
similarity_score = fasttext_model.wv.similarity("good", "bad")
print("Similarity Score is : ", similarity_score)
```

```python
similarity_score = fasttext_model.wv.similarity("good", "great")
print("Similarity Score is : ", similarity_score)
```
